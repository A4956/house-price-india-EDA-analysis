```python
# The average, median, and standard deviation of house prices:-

average_price=df['Price'].mean()
median_price=df['Price'].median()
st_dev_price=df['Price'].std()

print("average price of house is:", average_price)
print("Median price of house is: ", median_price)
print("st deviation price of house is :", st_dev_price)
```

Interpretaion- The average house price is around 5.39 lakhs which is noticeably higher than median price 4.5 lakhs. Since mean value is gretaer than meadian (mean > median), ity defines that there are preseence of outliners in price which pulling the averageup- so its a sigh of right skewed.

```python
#number of most common bedrooms is:-

most_common_bedroom=df['number of bedrooms'].mode()[0]


print("Most common number of bedrooms:" ,most_common_bedroom)
```

Interpretation- 3 bedroom house are the most standard to offer houses. Buyers may prefer 3 bedroom homes as balance between affordability and space. Developer can build more than 3 bedroom house depends on the demand.

```python
skew_value=df["Price"].skew()
print("Skewness of price:", skew_value )

meanprice=df['Price'].mean()
median_price=df['Price'].median()
mode_price=df['Price'].mode()[0]

print("mean price: ", meanprice)
print("median_price: ", median_price)
print("Mode_price: ", mode_price)

plt.hist(df['Price'], bins=30,color="skyblue",edgecolor= "darkblue")
plt.title("Distribution of house price to check skewness")
plt.xlabel("Price")
plt.ylabel("frequency")
plt.show()
```

Interpretaion- Skewness of price: 4.2 .Histogram is showing that the price of houses are right skewed means there are some houses price that are higer than the rest. mean> median>mode. Means there are few expensive properties that are raising the average.

```python
# Average price for each number of bedrooms-

bedroom_group = df.groupby("number of bedrooms")["Price"].mean().round().sort_index()

print(bedroom_group)
```

Interpretation -The mean or average price of bedroom was steady till 1 bedroom(around 3.89 lakhs) to 8 bedroom(around 8.2 lakh) are intutive. but 9 , 10 and 11 bedroom price is little unreliable. 33 bedroom house price is also unreliable compared to others.

```python
# relationship between living area and price

area_col = 'Area of the house(excluding basement)'
correlation = df[area_col].corr(df['Price'])
print(f"Correlation between '{area_col}' and Price: {correlation:.3f}")

plt.scatter(df[area_col], df['Price'], alpha=0.3,color='darkgreen', s=10)
plt.xlabel('Area of the house(excluding basement')
plt.ylabel('Price')
plt.title('living area (without basement) vs Price')
```

Interpretation- 0.615 correlation is positive relationship.larger living area is associated with higher price but the relation ship might be affected by grade, location(pincode), condition , waterfront. This condition clearly driven the price . Since two houses of similar area can have very different prices.

```python
# Anomalies in houses with high prices but low areas

Q75_price =df['Price'].quantile(0.75)
Q25_area=df['Area of the house(excluding basement)'].quantile(0.25)

print("price greater than 75 percentile:" , Q75_price)
print("and area less than 25 percentile:", Q25_area)

#Anomalies
anomalies = df[(df['Price'] > Q75_price) & (df[area_col] < Q25_area)]

print("Number of anomalie houses:" ,{len(anomalies)})

anomalies[['id', 'Area of the house(excluding basement)', 'number of bedrooms', 'living area', 'Price']].sort_values('Price', ascending=False).head(5)
```

139 anomalous properties (houses exhibiting unusual pricing structures). These homes represent the intersection of two extremes: they fall into the highest 25% price tier (above ₹645,000), yet occupy the smallest 25% size(1200 sq ft) bracket.This indicates a highly localized premium, where buyers are paying top-tier prices for compact living spaces."

```python
#Comparing average price based on number of floors and houses with waterfront.

avg_price_floors= df.groupby('number of floors')['Price'].mean().round()
avg_price_with_waterfront= df.groupby('waterfront present')['Price'].mean().round()

print('average price of houses based on floors: \n', avg_price_floors)
print('Average number of floor with water front : \n', avg_price_with_waterfront)

by_both=df.groupby(['number of floors', 'waterfront present'])['Price'].mean().head().round()
print("average price by floors with water fall: \n", by_both)
```

waterfront status is a much more powerful driver of price than the number of floors. A 1-story house with a water view is worth twice as much as a 2-story house without one.

```python
#minimum and maximum house price
min_houseprice= df['Price'].min()
max_houseprice= df['Price'].max()

print("minimum house price: \n", min_houseprice)
print("Maximum house price: \n", max_houseprice)
```
